```python
import json
```

```python
def load_data(filename): #function to load JSON data and make it a Python list
    with open(filename,"r") as f:
        data=json.load(f)
    return data
```

```python
datav1=load_data("customerfeedback.json")
```

```python
#ISSUES: data is inconsistent; we have mixed types; missing values and duplicates 
```

```python
def clean_data(data):
    text_to_num={"one":1.0,"two":2.0,"three":3.0,"four":4.0,"five":5.0}
    unique_users=set()
    cleaned_data=[]

    for user in data: 
        raw_rating=str(user["rating"]).strip().lower() #remove trailing spaces
        if raw_rating in text_to_num:
            user["rating"]=text_to_num[raw_rating]#converting alphabetical ratings to numerical ratings 
        user['rating'] = raw_rating


        raw_age= user.get("age") #if no age is given, set age as None
        if (raw_age)==None:
            user["age"]=None
            
         #handle duplicate values - deduplication
        if(user['name'].strip() in unique_users):
            continue
        unique_users.add(user['name'])
        cleaned_data.append(user)

    return cleaned_data 
```

```python
datav2=clean_data(datav1)
print(datav2)
```

```python
#Get the average rating and poor rating percentage from all user ratings
def get_insights(data):
    total_rating=0
    for user in data:
        total_rating+= float(user["rating"])
    poor_rating=0
    for user in data:
        if float(user["rating"])<3:
            poor_rating+=1
            
    print(f"Average rating is : {total_rating/len(data)}")
    print(f"Poor rating percentage is : {poor_rating/len(data)*100}%")
        
        
```

```python
get_insights(datav2)
```

```python
#product recommendation feature - RULE BASED
def get_recommendations(data):
    recommendations = []
    
    for user in data:
        curr_recomm = {}
        curr_recomm['name'] = user['name']
        if(float(user['rating']) >= 4):
            curr_recomm['brand'] = "Apple"
        else:
            curr_recomm['brand'] = "Samsung"

        recommendations.append(curr_recomm)
    
    return recommendations
```

```python
recommendations=get_recommendations(datav2)
print(recommendations)
```
