```python
#goal: clean a raw dataset and export the result to a new csv
# covers: removing duplicates, sorting, filling nulls,
# resetting index, and saving output as csv
# dataset: raw_data.csv
# output: sorted_data.csv
```

```python
import pandas as pd
```

```python
df = pd.read_csv("raw_data.csv")
```

```python
df2=df.copy()
df2.drop_duplicates(inplace=True)
df2.sort_values("income", inplace=True)
df2=df2.fillna(0)
df2=df2.reset_index(drop=True)
df2.to_csv("sorted_data.csv")
```
