```python
import pandas as pd
```

```python
df = pd.read_csv("raw_data.csv")
```

```python
df2=df.copy()
```

```python
# Transformation / Feature Engineering
df2 = df.copy()

# apply a function to a row or col using lambda
df2["tax"] = df2["income"].apply(lambda x : "10%" if x < 60000 else "20%")
df2
```

```python
#mapping out values 
new_gender_vals = {"Male": "M", "Female": "F", "Unknown": "U"}
df2["gender"] = df2["gender"].map(new_gender_vals)
df2
```

```python
# assign some new values
df2=df2.assign(new_income = df2["income"] * 1.1)
df2
```

```python
# replace specific values
df2['country'] = df2["country"].replace("USA", "US")
df2['name'] = df2['name'].replace("Alex", "Alex Hales")
df2
```

```python
# Transformation - Sorting
df2 = df.copy()

# Renaming
df2.columns = ["Id", "Name", "Age", "Country", "Gender", "Income"]
df2.rename(columns={"Name": "Initials"}, inplace=True) #for changing column label
df2.rename(index={0: "first"},inplace=True)
df2#for changing row label 
```

```python
# Sorting - values & index                  # sort values in ascending
sorted_df2=df2.sort_values(["Age", "Income"])            # sorts for age, if age is same then sorts for income 
sorted_df2
```

```python
 # Reset Index
sorted_df2.reset_index()  
```

```python
# Ranking
df2["Ranking"] = df2["Income"].rank(ascending=False,method="dense")
df2
```
