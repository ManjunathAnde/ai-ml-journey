```python
import pandas as pd
```

```python
df = pd.read_csv("raw_data.csv")
```

```python
df
```

```python
df.isnull().sum()
```

```python
'''Whenever the sum of missing values is high, it is not advisable to use that column 
for major analysis or model training'''
```

```python
df.dropna() #empty col drop {atleast one}
df.dropna(axis=1) # empty row drop {atleast one}

df.fillna("missing") #fills  a specific thing in missing 
```

```python
cleaned_data = df.copy() #creating a copy of the original data and keeping the mean age in missing fields
age_mean=cleaned_data["age"].mean()
cleaned_data["age"].fillna(age_mean)
```

```python
cleaned_data.ffill() #uses above value in missing fields in up to down direction
cleaned_data.bfill()  #uses above value in missing fields in down to up direction
```
