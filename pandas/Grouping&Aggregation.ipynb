```python

# Grouping and Aggregation 

# goal: learn how to group data by a category and run
# calculations on each group separately
# Concepts:
# groupby() to split data into groups
# aggregation functions - mean, min, max
# agg() for multiple aggregations at once
# enaming aggregated columns
# grouping and aggregating across multiple columns
# quick scatter plot to spot a pattern
```

```python
import pandas as pd
```

```python
df = pd.read_csv("raw_data.csv")
```

```python
df.groupby('country')["income"].mean()
```

```python
df.groupby('country')["income"].min() #Categorization, then aggregation 
```

```python
df.groupby('gender')["income"].max()
```

```python
df.groupby("country")["income"].agg(["mean","min","max"]) #getting aggregates by grouping incomes of countries 
```

```python
df.groupby("country")["income"].agg(avg="mean",min_salary="min",max_salary="max") #renaming cols
```

```python
df.groupby("country").agg({
    "age": "max",
    "income": "mean"
})   # aggregate on multiple cols

```

```python
df.groupby("country").agg(
    avg_age=("age", "mean"),
    avg_salary=("income", "mean")
)   # rename aggregates on multiple cols
```

```python
df.plot(kind="scatter",x="age",y="income")
```
