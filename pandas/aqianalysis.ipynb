```python
# DATA: globalAirQuality.csv (Kaggle)
# trying to understand how to work with real-world data
# using a global air quality dataset from Kaggle 

# CONCEPTS COVERED:
# Loading and exploring CSV data
# loc vs iloc (label vs index-based access)
# Filtering with conditions
# Query method for dynamic filtering
```

```python
df = pd.read_csv("globalAirQuality.csv")
```

```python
df[["city","aqi"]]
```

```python
# Find the row with the highest AQI
max_aqi_row = df.loc[df['aqi'].idxmax()]

# Print just the city and its AQI
print(f"The city with the highest AQI is {max_aqi_row['city']} with an AQI of {max_aqi_row['aqi']}")
```

```python
df.describe()
```

```python
df.loc[0:2] #location based acces of data
```

```python
df.iloc[2:5,3:6] #index based access [end excluded] 
```

```python
#Using loc or iloc returns a VIEW of data, not a copy. So, any changes will disturb the modify data
```

```python
'''Filtering of data'''
```

```python
df[ (df['aqi']>100) & (df['temperature'] <50) ]
```

```python
df[df['aqi']>100]['city'] #The labels from original data are preserved even when filters are applied 
```

```python
"""Query Rules"""
val= int(input("Please enter an AQI. We will show cities with better then the AQI enetered"))
df.query("aqi>@val & temperature >30")[["city","country","aqi"]]
```
